### function的呼叫方式

### 引數值的呼叫


```python
def menu(wine,entree,dessert):
    print(f"前酒:{wine}")
    print(f"主餐:{entree}")
    print(f"甜點:{dessert}")
```

```python
#引數值的呼叫
menu('白酒','牛排','蛋糕')
```

### 引數名稱的呼叫
- 使用參數的名稱當作引數名稱
- 可以不依照參數順序

```python
def menu(wine,entree,dessert):
    print(f"前酒:{wine}")
    print(f"主餐:{entree}")
    print(f"甜點:{dessert}")
```

```python
menu(wine='白蘭地',entree='牛排',dessert='蛋糕')
```

```python
menu(dessert='蛋糕',wine='白蘭地',entree='牛排')
```
