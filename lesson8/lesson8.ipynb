### tuple
- immutable(不可變的變數)
- sequence(串列資料)
- iterable(可重複讀取)
- 拆解法
- 用括號符號()建立
- typehint(tuple)

```python
#拆解法
name, age = "Winnie", 21
print(name)
print(age)
```

### list
- mutable(可變的變數)
- sequence(串列資料)
- iterable(可重複讀取)
- 拆解法
- 用中括號符號[]建立
- typehint(list)

```python
#建立list
weekdays = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
type(weekdays)
```

```python
#取出元素和修改元素
#使用[index]
#第一個項目索引編號是0
#而倒數的話第一個項目則是看有幾個項目，比如有7個項目的話則會是-7
weekdays[0]
weekdays[5]
weekdays[-3]
```

```python
#[slice]
#[起始(包含):終點(不包含)]

weekdays[0:3]
weekdays[-5:-1]
weekdays[-3:]#取最後三個
weekdays[:3]#取前三個
```

```python
for day in weekdays:
    print(day)
```

### dict
- mutable(可變的變數)
- mapping(mapping資料)
- iterable(可重複讀取)
- 拆解法
- 用大括號符號{key:value}建立
- typehint(dict)

```python
codes = {"tw":"Taiwan", "jp":"Japan", "hk":"Hong Kong"}
codes
```

```python
codes["tw"]
```

```python
for key in codes:
    print(key)
    print(codes[key])
    print("===============")
```
